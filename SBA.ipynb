```python
#importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('woocommerce-product-export.csv') #Reading Data from the CSV file

df.info() #Showing a concise summary using the info() method
```

```python
df.head(5) # print the first 5 rows
```

```python
df.tail(5) # Print the last five rows
```

```python
df[['total_profit', 'month_number']] #Print  the "total_profit" and “month_number” columns only
```

```python
plt.bar('month_number' , 'total_profit', color='blue', width=0.5, data=df)
plt.title('Company Profit Per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')

plt.show()
```

```python
#Read the total Profit and show it using the bar plot
plt.bar('month_number', 'total_profit', color='blue', width=0.75, data=df)

plt.grid( linestyle='--', linewidth=1)
plt.title('Company Profit Per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')

plt.show()
```

```python
#Read the total Profit and show it using the bar plot
plt.plot(
    'month_number', 
    'total_profit', 
    color='red', 
    linestyle='--',
    marker='o', 
    markerfacecolor='black', 
    linewidth=3, 
    data=df
    )

#plt.grid( linestyle='--', linewidth=1)
plt.title('Company Profit Per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')

plt.legend(loc='lower right')

plt.show()
```
